```python
not df["Postal Code"].astype(str).str.contains(r"[A-Za-z]", regex=True, na=False).any()
```

```python
import numpy as np

# Identify invalid postal codes
invalid = ~df["Postal Code"].astype(str).str.fullmatch(r"\d+", na=False)

# Replace invalid values with NaN
df.loc[invalid, "Postal Code"] = np.nan

# Replace NaN with mode
df["Postal Code"] = df["Postal Code"].fillna(df["Postal Code"].mode()[0])
```
